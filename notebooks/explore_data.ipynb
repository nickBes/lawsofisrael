# Quick data exploration

A small sample of the Israeli bills dataset.

```python
from pathlib import Path

import pandas as pd
from IPython.display import HTML
from openpyxl import load_workbook

data_path = next(Path("../dataset").glob("*.xlsx"))
df = pd.read_excel(data_path)
print(f"{data_path.name}: {df.shape[0]:,} rows x {df.shape[1]} columns")
df.head()
```

```python
df.info()
df.sample(5, random_state=42)
```

```python
sheet = load_workbook(data_path, data_only=False).active
link_column = next(cell.column for cell in sheet[1] if cell.value == "קישור")
link_formulas = pd.Series(
    [sheet.cell(row, link_column).value for row in range(2, sheet.max_row + 1)],
    index=df.index,
)
urls = link_formulas.str.extract(r'(https?://[^"]+)', expand=False)

link_sample = df.loc[urls.notna(), ["שם", "סטטוס"]].assign(url=urls.dropna())
HTML(link_sample.sample(5, random_state=42).to_html(render_links=True))
```
